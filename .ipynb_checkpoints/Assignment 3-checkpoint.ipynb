```python
# ==========================================
# Q1. Nested Dictionary - Product Management
# ==========================================

products = {
    101: {"name": "Laptop", "category": "Electronics", "price": 1200, "stock": 50},
    102: {"name": "Shirt", "category": "Apparel", "price": 25, "stock": 200},
    103: {"name": "Coffee Maker", "category": "Home Appliances", "price": 80, "stock": 30}
}

# Increase shirt stock to 250
products[102]["stock"] = 250
print("Updated Products:")
print(products)

# Add a new product
products[104] = {
    "name": "Smartphone",
    "category": "Electronics",
    "price": 800,
    "stock": 100
}

print("\nProducts After Adding Smartphone:")
print(products)
```

```python
# ==========================================
# Q2. Age Group Classification
# ==========================================

age = int(input("Enter your age: "))

if age < 0:
    print("Invalid age")
elif age <= 12:
    print("Child")
elif age <= 19:
    print("Teenager")
elif age <= 64:
    print("Adult")
else:
    print("Senior")
```

```python
# ==========================================
# Q3. Store and Display Book Details in Tuple
# ==========================================

title = input("Enter the title of the book: ")
author = input("Enter the author's name: ")
year = int(input("Enter the year of publication: "))
price = float(input("Enter the price of the book: "))

book = (title, author, year, price)

print("\nBook Details:")
print("Title:", book[0])
print("Author:", book[1])
print("Year of Publication:", book[2])
print("Price: $", book[3], sep="")

```

```python
# ==========================================
# Q4. Positive, Negative, or Zero
# ==========================================

number = float(input("Enter a number: "))

if number > 0:
    print("The number is Positive")
elif number < 0:
    print("The number is Negative")
else:
    print("The number is Zero")
```

```python
# ==========================================
# Q5. Check Whether a Value Exists in a Tuple
# ==========================================

my_tuple = (5, 10, 15, 20, 25)

number = int(input("Enter an integer: "))

if number in my_tuple:
    print("The number exists in the tuple")
else:
    print("The number does not exist in the tuple")
```

```python
# ==========================================
# Q6. Remove Duplicates Using a Set
# ==========================================

shopping_list = ["apple", "banana", "apple", "orange", "banana", "grape"]

# Convert list to set, then convert back to list
shopping_list = list(set(shopping_list))

print("List after removing duplicates:")
print(shopping_list)
```

```python
# ==========================================
# Q7. Student Details - Add and Remove
# ==========================================

student_details = {
    "Ali": {"age": 20, "grade": "A", "city": "Karachi"},
    "Sara": {"age": 21, "grade": "B", "city": "Lahore"},
    "Omar": {"age": 19, "grade": "A", "city": "Islamabad"},
    "Hina": {"age": 22, "grade": "C", "city": "Faisalabad"}
}

# Add a new student using update()
student_details.update({
    "Zara": {"age": 20, "grade": "B", "city": "Multan"}
})

# Remove Omar using pop()
student_details.pop("Omar")

print("Updated Student Details:")
print(student_details)
```

```python
# ==========================================
# Q8. Book Inventory Management
# ==========================================

inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5},
    "Rich dad poor dad": {"price": 8.99, "quantity": 8},
    "Harry Potter": {"price": 12.50, "quantity": 2}
}

# Add a new book using update()
inventory.update({
    "Atomic Habits": {"price": 15.99, "quantity": 0}
})

# Get all book titles using keys()
print("All Book Titles:")
print(list(inventory.keys()))

# Remove books that are out of stock
for book_title in list(inventory.keys()):
    if inventory[book_title]["quantity"] == 0:
        inventory.pop(book_title)

print("\nInventory After Removing Out-of-Stock Books:")
print(inventory)
```

```python
# ==========================================
# Q9. Print All Dictionary Items
# ==========================================

d = {"x": 1, "y": 2}

# Use items() to display key-value pairs
print("Dictionary Items:")
for key, value in d.items():
    print(key, ":", value)
```

```python
# ==========================================
# Q10. Vowel, Consonant, or Invalid Input
# ==========================================

letter = input("Enter a single letter: ")

if len(letter) == 1 and letter.isalpha():
    if letter.lower() in "aeiou":
        print(letter, "is a Vowel")
    else:
        print(letter, "is a Consonant")
else:
    print("Please enter valid alphabet")
```
