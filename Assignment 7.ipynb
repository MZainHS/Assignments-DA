```python
# ==========================================
# QUESTION 1: EMPLOYEE DATA BANK (USING OOP)
# ==========================================

# Creating Employee Class

class Employee:

    # Constructor to initialize employee details
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):

        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    # Method to display employee information
    def describe_employee(self):

        print("\n----- Employee Details -----")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)

    # Method to update employee experience
    def update_year_of_experience(self, new_experience):

        self.year_of_experience = new_experience
        print("Experience updated successfully!")

    # Method to increase employee salary
    def increase_salary(self, increment):

        self.salary += increment
        print("Salary increased successfully!")


# Creating Employee 101
employee101 = Employee(
    "Ali", 101, "Karachi", 50000, "Manager",
    2020, "MBA", 5
)

# Printing four attributes individually
print("Employee Name:", employee101.name)
print("Employee Number:", employee101.number)
print("Employee Address:", employee101.address)
print("Employee Salary:", employee101.salary)

# Calling all methods
employee101.describe_employee()

employee101.increase_salary(10000)
employee101.update_year_of_experience(6)

# Displaying updated information
employee101.describe_employee()


# Creating 10 Different Employee Instances

employee102 = Employee("Sara", 102, "Lahore", 45000, "Accountant", 2021, "B.Com", 4)

employee103 = Employee("Ahmed", 103, "Islamabad", 60000, "Developer", 2019, "BSCS", 6)

employee104 = Employee("Zara", 104, "Karachi", 55000, "HR Officer", 2022, "MBA", 3)

employee105 = Employee("Usman", 105, "Multan", 40000, "Sales Executive", 2023, "BBA", 2)

employee106 = Employee("Ayesha", 106, "Faisalabad", 70000, "Team Lead", 2018, "MSCS", 7)

employee107 = Employee("Hamza", 107, "Peshawar", 35000, "Data Analyst", 2022, "BSIT", 3)

employee108 = Employee("Fatima", 108, "Quetta", 48000, "Designer", 2021, "BS Design", 4)

employee109 = Employee("Bilal", 109, "Hyderabad", 52000, "Supervisor", 2020, "BBA", 5)

employee110 = Employee("Hina", 110, "Sialkot", 65000, "Project Manager", 2017, "MBA", 8)


# Storing all employees in a list
employees = [
    employee101, employee102, employee103, employee104,
    employee105, employee106, employee107, employee108,
    employee109, employee110
]


# Displaying details of all 10 employees
print("\n========== ALL EMPLOYEE RECORDS ==========")

for employee in employees:
    employee.describe_employee()


# Updating Employee Information
print("\n========== EMPLOYEE UPDATES ==========")

employee102.increase_salary(5000)
employee103.increase_salary(8000)
employee104.update_year_of_experience(4)
employee105.update_year_of_experience(3)
employee106.increase_salary(10000)

# Displaying updated employee records
employee102.describe_employee()
employee103.describe_employee()
employee104.describe_employee()
employee105.describe_employee()
employee106.describe_employee()
```
